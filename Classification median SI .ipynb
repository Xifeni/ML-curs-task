Выводы: Задача классификации SI

Классификация индекса селективности оказалась наиболее сложной задачей в рамках исследования. Тем не менее, переход к классификации позволил преодолеть порог в 70% точности, что значительно превосходит результаты, полученные ранее в задачах регрессии.

1. RandomForest показал лучший результат в этой задаче. Модель достигла accuracy 0.706 и самой высокой precision 0.708, используя оптимальный набор из 25 признаков. В условиях высокой зашумленности данных SI (вычисляемы признак от IC50 и CC50, наследуют неточность от обоих), RF проявил себя как наиболее стабильный алгоритм, способный эффективно усреднять ошибки отдельных деревьев.

2. CatBoost продемонстрировал практически идентичную точность accuracy 0.701 при использовании всего 25 признаков. Незначительное отставание от RF по метрике precision компенсируется стабильностью на кросс-валидации CV F1: 0.66. Это подтверждает, что для данной задачи древесные ансамбли являются единственным эффективным инструментом, способным выявить скрытые закономерности.

3. Logistic Regression и SVM показали заметно более низкие результаты. Это указывает на то, что селективность соединений определяется сложным сочетанием признаков, которое невозможно описать линейно или через простые геометрические расстояния. Метрики этих моделей подтверждают отсутствие явных линейных трендов в распределении индекса селективности.

4. XGBoost accuracy 0.67 неожиданно отстал от других ансамблей. Несмотря на использование всего 15 признаков, модель оказалась менее эффективной в обобщении данных SI по сравнению с CatBoost и RF, что может быть связано с особенностями структуры шума в этой конкретной выборке.
